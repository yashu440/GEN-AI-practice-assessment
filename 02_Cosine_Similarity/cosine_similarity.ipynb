```
# ============================================================
# COSINE SIMILARITY MATHEMATICALLY
# ============================================================

import numpy as np

print("=" * 60)
print("COSINE SIMILARITY")
print("=" * 60)


# ------------------------------------------------------------
# 1. Define two vectors
# ------------------------------------------------------------

vector_A = np.array([1, 2, 3])
vector_B = np.array([4, 5, 6])

print("\nVector A:")
print(vector_A)

print("\nVector B:")
print(vector_B)


# ------------------------------------------------------------
# 2. Calculate dot product
# ------------------------------------------------------------

dot_product = np.dot(vector_A, vector_B)

print("\n" + "-" * 60)
print("STEP 1: DOT PRODUCT")
print("-" * 60)

print("A · B =", dot_product)


# ------------------------------------------------------------
# 3. Calculate magnitude of Vector A
# ------------------------------------------------------------

magnitude_A = np.linalg.norm(vector_A)

print("\n" + "-" * 60)
print("STEP 2: MAGNITUDE OF VECTOR A")
print("-" * 60)

print("||A|| =", magnitude_A)


# ------------------------------------------------------------
# 4. Calculate magnitude of Vector B
# ------------------------------------------------------------

magnitude_B = np.linalg.norm(vector_B)

print("\n" + "-" * 60)
print("STEP 3: MAGNITUDE OF VECTOR B")
print("-" * 60)

print("||B|| =", magnitude_B)


# ------------------------------------------------------------
# 5. Calculate denominator
# ------------------------------------------------------------

denominator = magnitude_A * magnitude_B

print("\n" + "-" * 60)
print("STEP 4: DENOMINATOR")
print("-" * 60)

print("||A|| × ||B|| =", denominator)


# ------------------------------------------------------------
# 6. Calculate Cosine Similarity
# ------------------------------------------------------------

cosine_similarity = dot_product / denominator

print("\n" + "-" * 60)
print("STEP 5: COSINE SIMILARITY")
print("-" * 60)

print("Cosine Similarity =", cosine_similarity)


# ------------------------------------------------------------
# 7. Calculate angle between vectors
# ------------------------------------------------------------

angle_radians = np.arccos(cosine_similarity)
angle_degrees = np.degrees(angle_radians)

print("\n" + "-" * 60)
print("ANGLE BETWEEN VECTORS")
print("-" * 60)

print("Angle in radians =", angle_radians)
print("Angle in degrees =", angle_degrees)


# ------------------------------------------------------------
# 8. Verify using sklearn
# ------------------------------------------------------------

from sklearn.metrics.pairwise import cosine_similarity

A = vector_A.reshape(1, -1)
B = vector_B.reshape(1, -1)

sklearn_result = cosine_similarity(A, B)[0][0]

print("\n" + "-" * 60)
print("VERIFICATION USING SCIKIT-LEARN")
print("-" * 60)

print("Cosine Similarity using sklearn =", sklearn_result)


# ------------------------------------------------------------
# 9. Test another pair of vectors
# ------------------------------------------------------------

vector_C = np.array([1, 0, 0])
vector_D = np.array([0, 1, 0])

manual_result = np.dot(vector_C, vector_D) / (
    np.linalg.norm(vector_C) * np.linalg.norm(vector_D)
)

print("\n" + "-" * 60)
print("SECOND EXAMPLE")
print("-" * 60)

print("Vector C:", vector_C)
print("Vector D:", vector_D)

print("Cosine Similarity =", manual_result)


# ------------------------------------------------------------
# 10. Test identical vectors
# ------------------------------------------------------------

vector_E = np.array([2, 4, 6])
vector_F = np.array([2, 4, 6])

identical_result = np.dot(vector_E, vector_F) / (
    np.linalg.norm(vector_E) * np.linalg.norm(vector_F)
)

print("\n" + "-" * 60)
print("THIRD EXAMPLE - IDENTICAL VECTORS")
print("-" * 60)

print("Vector E:", vector_E)
print("Vector F:", vector_F)

print("Cosine Similarity =", identical_result)


# ------------------------------------------------------------
# 11. Test opposite vectors
# ------------------------------------------------------------

vector_G = np.array([1, 2, 3])
vector_H = np.array([-1, -2, -3])

opposite_result = np.dot(vector_G, vector_H) / (
    np.linalg.norm(vector_G) * np.linalg.norm(vector_H)
)

print("\n" + "-" * 60)
print("FOURTH EXAMPLE - OPPOSITE VECTORS")
print("-" * 60)

print("Vector G:", vector_G)
print("Vector H:", vector_H)

print("Cosine Similarity =", opposite_result)


# ------------------------------------------------------------
# 12. Interpretation
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("INTERPRETATION")
print("=" * 60)

print("""
Cosine similarity measures the similarity between two vectors.

Value close to +1  -> Very similar
Value close to  0  -> Not similar / perpendicular
Value close to -1  -> Opposite direction

The formula is:

Cosine Similarity = (A · B) / (||A|| × ||B||)
""")


# ------------------------------------------------------------
# 13. Final result
# ------------------------------------------------------------

print("=" * 60)
print("COSINE SIMILARITY PROGRAM COMPLETED SUCCESSFULLY")
print("=" * 60)
```
